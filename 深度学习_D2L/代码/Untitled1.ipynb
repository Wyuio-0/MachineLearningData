## 2.4.1 导数和微分 (Derivatives and Differentiation)
在深度学习中，我们通过最小化损失函数来优化模型。
导数（Derivative）表示函数在某一点处的**瞬时变化率**，在几何上对应曲线在该点切线的**斜率**：
$$f'(x) = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}$$

```python
# ==============================================================================
# 【代码 2.4.1-1】导入依赖库与定义实验函数 f(x)
# 功能：引入 numpy、matplotlib_inline 以及 d2l 工具包，定义二次函数 f(x) = 3x^2 - 4x
# ==============================================================================
%matplotlib inline
import numpy as np
from matplotlib_inline import backend_inline
from d2l import torch as d2l

def f(x):
    return 3 * x ** 2 - 4 * x
# ------------------------------------------------------------------------------
# 【数学解析】理论导数 f'(x) = 6x - 4。在 x = 1 处，理论导数值 f'(1) = 6(1) - 4 = 2
# ==============================================================================
```

### 数值极限逼近验证 (Numerical Differentiation)
通过令 $h$ 逐步缩小（0.1, 0.01, 0.001...），验证有限差分 $\frac{f(x+h) - f(x)}{h}$ 是否向理论导数值 2 收敛。

```python
# ==============================================================================
# 【代码 2.4.1-2】数值导数逼近计算
# 功能：计算有限差分差商，观察步长 h 逐渐缩小时差商的变化趋势
# ==============================================================================
def numerical_lim(f, x, h):
    return (f(x + h) - f(x)) / h

h = 0.1
for i in range(5):
    print(f'h={h:.5f}, numerical limit={numerical_lim(f, 1, h):.5f}')
    h *= 0.1
# ------------------------------------------------------------------------------
# 【结果解读】随着 h 趋于 0，数值极限高度逼近 2.0，证明 x = 1 处瞬时变化率就是 2
# ==============================================================================
```

### 导数的几何意义可视化（曲线与切线）
在 $x = 1$ 处：$f(1) = 3(1)^2 - 4(1) = -1$。
切线斜率 $k = f'(1) = 2$。
切线点斜式方程为：$y - (-1) = 2(x - 1) \implies y = 2x - 3$。
使用 `d2l.plot` 将函数曲线与切线绘制在同一画板中进行直观对比。

```python
# ==============================================================================
# 【代码 2.4.1-3】绘制函数曲线与在 x = 1 处的切线
# 功能：使用 d2l.plot 将 f(x) 与切线 y = 2x - 3 叠加绘制
# ==============================================================================
x = np.arange(0, 3, 0.1)
d2l.plot(x, [f(x), 2 * x - 3], 'x', 'f(x)', legend=['f(x)', 'Tangent line (x=1)'])
# ------------------------------------------------------------------------------
# 【可视化解读】实线为抛物线 f(x)，虚线为在 x=1 点的切线，直观展现切线斜率即导数
# ==============================================================================
```
