```python
# 【所属小节】3.1.2 矢量化加速 - 导入必要工具库
# 【作用说明】导入 math、time、numpy、torch 以及 d2l 绘图工具包
%matplotlib inline
import math
import time
import numpy as np
import torch
from d2l import torch as d2l
```

```python
# 【所属小节】3.1.2 矢量化加速 - 实例化测试向量
# 【作用说明】实例化两个全为 1 的 10000 维向量，用于对比 Python 原生 for 循环与张量矢量化加法的性能
n = 10000
a = torch.ones([n])
b = torch.ones([n])
```

```python
# 【所属小节】3.1.2 矢量化加速 - 定义高精度计时器
# 【作用说明】封装 Timer 类记录代码块的多次运行时间、总时间、平均时间与累计时间，用于后续基准测试
class Timer:
    #@save
    """记录多次运行时间"""
    def __init__(self):
        self.times = []
        self.start()

    def start(self):
        """启动计时器"""
        self.tik = time.time()

    def stop(self):
        """停止计时器并将时间记录在列表中"""
        self.times.append(time.time() - self.tik)
        return self.times[-1]

    def avg(self):
        """返回平均时间"""
        return sum(self.times) / len(self.times)

    def sum(self):
        """返回时间总和"""
        return sum(self.times)

    def cumsum(self):
        """返回累计时间"""
        return np.array(self.times).cumsum().tolist()
```

```python
# 【所属小节】3.1.2 矢量化加速 - 基准测试 1: Python for 循环逐元素相加
# 【作用说明】使用 Python 原生循环进行 10000 次标量加法，观察运行耗时
c = torch.zeros(n)
timer = Timer()
for i in range(n):
    c[i] = a[i] + b[i]
f'{timer.stop():.5f} sec'
```

```python
# 【所属小节】3.1.2 矢量化加速 - 基准测试 2: 张量重载 + 运算符矢量化相加
# 【作用说明】使用底层 C++/CUDA 优化的矢量化操作直接相加，观察数十倍至数百倍的数量级加速
timer.start()
d = a + b
f'{timer.stop():.5f} sec'
```

```python
# 【所属小节】3.1.3 正态分布与平方损失 - 正态分布概率密度函数定义
# 【作用说明】根据高斯分布数学公式 p(x) = (1 / sqrt(2*pi*sigma^2)) * exp(-0.5/sigma^2 * (x - mu)^2) 计算概率密度
def normal(x, mu, sigma):
    p = 1 / math.sqrt(2 * math.pi * sigma**2)
    return p * np.exp(-0.5 / sigma**2 * (x - mu)**2)
```

```python
# 【所属小节】3.1.3 正态分布与平方损失 - 正态分布曲线可视化
# 【作用说明】改变均值 mu(控制横轴对称中心偏移)与标准差 sigma(控制离散平坦程度)，使用 d2l.plot 绘制对比曲线
x = np.arange(-7, 7, 0.01)
# 均值和标准差对
params = [(0, 1), (0, 2), (3, 1)]
d2l.plot(x, [normal(x, mu, sigma) for mu, sigma in params], xlabel='x',
         ylabel='p(x)', figsize=(4.5, 2.5),
         legend=[f'mean {mu}, std {sigma}' for mu, sigma in params])
```

## 课后练习题代码实战与算法验证

包含第 3.1 节课后关于均值解析解、无偏置解析正规方程、以及拉普拉斯噪声（L1/中位数回归）的数值仿真。


```python
# 【课后练习 1】寻找常数 b 最小化平方误差 sum (xi - b)^2
# 【作用说明】随机生成正态分布数据，数值验证使误差和最小的最优 b 严格等于样本均值 mean(x)
torch.manual_seed(42)
data_x = torch.randn(100) * 2 + 5  # 均值为 5，方差为 4 的正态分布样本

# 解析解最优 b* = 样本均值
b_optimal = data_x.mean().item()

# 在最优值邻域内采样测试损失函数曲线
b_candidates = np.linspace(b_optimal - 2, b_optimal + 2, 50)
losses = [((data_x - b) ** 2).sum().item() for b in b_candidates]

print(f"数据样本均值 (解析最优解 b*): {b_optimal:.4f}")
print(f"网格搜索得到的最小损失处 b: {b_candidates[np.argmin(losses)]:.4f}")
```

```python
# 【课后练习 2】无偏置线性回归解析解与正规方程 w* = (X^T X)^(-1) X^T y 验证
# 【作用说明】生成合成线性数据，使用 PyTorch 矩阵求逆直接一步求解最优权重，并与真实权重对比
true_w = torch.tensor([1.5, -2.0, 3.2]).reshape(-1, 1)
X = torch.randn(500, 3)
y = torch.matmul(X, true_w) + torch.randn(500, 1) * 0.05  # 微弱高斯噪声

# 计算正规方程解析解 w* = (X^T X)^(-1) X^T y
w_analytical = torch.inverse(torch.matmul(X.T, X)).matmul(X.T).matmul(y)

print("真实权重 true_w:\n", true_w.squeeze())
print("正规方程求得的解析解 w_analytical:\n", w_analytical.squeeze())
print("绝对误差范数:", torch.norm(true_w - w_analytical).item())
```

```python
# 【课后练习 3】拉普拉斯噪声与次梯度下降(绝对值 L1 损失中位数回归)
# 【作用说明】当噪声服从对称指数分布(拉普拉斯)时，负对数似然对应 L1 损失。
# 演示采用衰减学习率解决驻点不可导处的震荡现象。
torch.manual_seed(42)
true_w_l1 = torch.tensor([2.5]).reshape(-1, 1)
X_l1 = torch.randn(300, 1)
# 生成拉普拉斯噪声 (通过指数分布合成)
laplace_noise = torch.distributions.Laplace(0, 1).sample((300, 1))
y_l1 = torch.matmul(X_l1, true_w_l1) + laplace_noise

# 使用次梯度下降训练 L1 损失
w_hat = torch.zeros(1, 1, requires_grad=True)
lr = 0.1

for epoch in range(1, 101):
    # L1 损失 (绝对值误差均值)
    loss = torch.mean(torch.abs(torch.matmul(X_l1, w_hat) - y_l1))
    loss.backward()
    
    with torch.no_grad():
        # 学习率动态衰减以避免最优点处振荡
        step_lr = lr / math.sqrt(epoch)
        w_hat -= step_lr * w_hat.grad
        w_hat.grad.zero_()

print(f"真实权重: {true_w_l1.item():.4f}, 训练估计值 w_hat: {w_hat.item():.4f}")
```
