# 2.2 数据预处理 (Data Preprocessing)
本 Notebook 包含使用 Pandas 读取真实数据、缺失值处理、虚拟变量独热编码与 PyTorch 张量转换的标准工业流。

## 2.2.1 读取数据集：创建与加载 CSV

```python
# 【所属小节】2.2.1 读取数据集
# 【作用说明】在 ../data 目录下创建微型房价示例 CSV 数据集（人工注入缺失值 NA）
import os

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny.csv')
with open(data_file, 'w', encoding='utf-8') as f:
    f.write('NumRooms,Alley,Price\n')  # 列名：房间数、巷子类型、房价
    f.write('NA,Pave,127500\n')        # 每行表示一个样本
    f.write('2,NA,106000\n')
    f.write('4,NA,178100\n')
    f.write('NA,NA,140000\n')
print('数据集创建完毕:', data_file)
```

```python
# 【所属小节】2.2.1 读取数据集
# 【作用说明】使用 pandas 读取刚刚创建的 CSV 文件，并打印结构
import pandas as pd

data = pd.read_csv(data_file)
print(data)
```

## 2.2.2 处理缺失值：特征拆分、均值填充与独热编码

```python
# 【所属小节】2.2.2 处理缺失值
# 【作用说明】使用 iloc 拆分输入特征 (前两列) 与预测目标 (最后一列)，并对数值型缺失值用列均值填充
inputs, outputs = data.iloc[:, 0:2], data.iloc[:, 2]
inputs = inputs.fillna(inputs.mean(numeric_only=True))
print('数值均值插值填充后的输入特征:\n', inputs)
print('\n预测目标标签:\n', outputs)
```

```python
# 【所属小节】2.2.2 处理缺失值
# 【作用说明】对离散文本特征进行独热编码 (One-Hot)，把 NaN 作为独立类别标识 (dummy_na=True)
inputs = pd.get_dummies(inputs, dummy_na=True, dtype=float)
print('独热编码转换后的特征矩阵:\n', inputs)
```

## 2.2.3 转换为张量格式

```python
# 【所属小节】2.2.3 转换为张量格式
# 【作用说明】将预处理完备的 Pandas 数据框转换为 PyTorch 深度学习所需的 Tensor 格式
import torch

X = torch.tensor(inputs.to_numpy(dtype=float))
y = torch.tensor(outputs.to_numpy(dtype=float))
print('特征张量 X:\n', X)
print('标签张量 y:\n', y)
```
