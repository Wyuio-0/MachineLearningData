## 2.1.2 运算符：按元素数学计算与张量拼接

```python
# 【所属小节】2.1.2 运算符
# 【作用说明】按元素（Element-wise）执行常见的加、减、乘、除、指数幂运算
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])
x + y, x - y, x * y, x / y, x ** y
```

## 2.1.5 节省内存与原地操作 (In-place Operations)

```python
# 【所属小节】2.1.5 节省内存
# 【作用说明】普通赋值 Y = Y + X 会分配新内存，导致前后内存地址 id(Y) 改变
before = id(Y)
Y = Y + X
id(Y) == before  # 结果为 False，分配了新内存
```

```python
# 【所属小节】2.1.5 节省内存
# 【作用说明】切片写入 Z[:] = X + Y 可以在预先分配好的内存中原地重写，内存地址不变
Z = torch.zeros_like(Y)
print('写入前 id(Z):', id(Z))
Z[:] = X + Y
print('写入后 id(Z):', id(Z))  # 地址完全相同！
```

```python
# 【所属小节】2.1.5 节省内存
# 【作用说明】运算符 X += Y 是原生原地操作（In-place），同样不产生内存重分配
before = id(X)
X += Y
id(X) == before  # 结果为 True
```
